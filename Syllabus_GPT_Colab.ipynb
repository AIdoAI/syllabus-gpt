## Step 2: Clone the Repository

## Step 3: Set Up API Keys

You'll need:
1. **OpenAI API Key** - Get from https://platform.openai.com/api-keys
2. **ngrok Auth Token** - Get from https://dashboard.ngrok.com/get-started/your-authtoken (free account)

````python
import os
from getpass import getpass

# Set OpenAI API key
openai_key = getpass("Enter your OpenAI API key: ")
os.environ['OPENAI_API_KEY'] = openai_key

# Create .env file
with open('.env', 'w') as f:
    f.write(f'OPENAI_API_KEY={openai_key}\n')

print("✅ OpenAI API key set!")
````

## Step 4: Set Up ngrok

ngrok creates a public URL that tunnels to your Colab instance.

````python
from pyngrok import ngrok, conf

# Set ngrok auth token
ngrok_token = getpass("Enter your ngrok auth token: ")
conf.get_default().auth_token = ngrok_token

print("✅ ngrok configured!")
````

## Step 5: Run Syllabus-GPT!

This cell will:
1. Start Streamlit in the background
2. Create a public ngrok URL
3. Display the URL for you to click

⚠️ **Keep this cell running!** If you stop it, the app will stop.

````python
import subprocess
import time
from pyngrok import ngrok

# Kill any existing Streamlit processes
!pkill -f streamlit

# Start Streamlit in background
print("🚀 Starting Streamlit...")
process = subprocess.Popen(
    ['streamlit', 'run', 'app.py', '--server.port', '8501'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

# Wait for Streamlit to start
time.sleep(5)

# Create ngrok tunnel
print("🌐 Creating public URL...")
public_url = ngrok.connect(8501)

print("\n" + "="*60)
print("✅ Syllabus-GPT is running!")
print("="*60)
print(f"\n🔗 Access your app here:\n{public_url}")
print("\n📸 Click the link above to open Syllabus-GPT!")
print("\n⚠️  Keep this cell running - stopping it will close the app")
print("="*60)
````

## 🐛 Troubleshooting

### "Module not found" errors
- Re-run Step 1 to reinstall packages

### ngrok URL doesn't work
- Check that you entered your ngrok auth token correctly
- Visit https://dashboard.ngrok.com/get-started/your-authtoken to get it again

### "OpenAI API key not found"
- Re-run Step 3 with your correct API key

### App is slow
- This is normal on Colab's free tier
- Processing PDFs takes 20-30 seconds

### Want to stop and restart?
- Run the stop cell
- Then re-run Step 5 to start again
